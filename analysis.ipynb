Loading packages and data from try_reduced.csv, stored in the data folder

```r
library(ggplot2)
library(dplyr)

dat <- read.csv("data/try_reduced.csv")
head(dat)
```

Creating a plot of seed dry mass for Fagus sylvatica

```r
dat |> filter(SpeciesName == "Fagus sylvatica",
              TraitName == "Seed dry mass") |>
  ggplot(aes(StdValue)) + geom_histogram() +
  scale_x_continuous(trans = "log10") +
  theme_light()
```
